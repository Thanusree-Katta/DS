```python
import pandas as pd
from scipy.stats import spearmanr
df=pd.DataFrame({
    'x':[111,222,333,444,5545],
    'y':[10,20,30,40,50]
})
corr_value, p_value=spearmanr(df['x'],df['y'])
print("Spearman correlation coefficient:",corr_value)
print("P-value:",p_value)
```

```python
def hamming_distance(str1, str2):
    if len(str1) != len(str2):
        raise ValueError("Strings must be of equal length")

    return sum(ch1!=ch2 for ch1, ch2 in zip(str1, str2))
#str1="Meghna is topper"
#str2="tansssis  topper"
str1="abc"
str2="def"
distance=hamming_distance(str1,str2)
print(f"Hamming distance between '{str1}' and '{str2}' is: {distance}")
```

```python
def lcs_length(x ,y):
    m, n = len(x), len(y)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m):
        for j in range(n):
            if x[i] == y[j]:
                dp[i + 1][j + 1] = dp[i][j] + 1
            else:
                dp[i + 1][j + 1] = max(dp[i + 1][j], dp[i][j + 1])
    return dp[m][n]

seq1 = "ABCDEF"
seq2 = "AEBDF"
length = lcs_length(seq1, seq2)
print(f"The length of the longest common subsequence between '{seq1}' and '{seq2}' is: {length}")
```
