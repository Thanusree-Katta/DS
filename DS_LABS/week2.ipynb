```python
import pandas as pd
import numpy as np
#Data cleaning and handling missing values
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print(df)
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print("original data:")
print(df)
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
```

```python
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print("original data:")
print(df)
df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print("original data:")
print(df)
df_drop_rows=df.dropna()
print("After dropping:\n",df_drop_rows)
```

```python
df_drop_cols=df.dropna(axis=1)
print("After dropping columns:\n",df_drop_cols)
```

```python
df_drop_rows=df.dropna(axis=0)
print("After dropping rows:\n",df_drop_rows)
```

```python
#removing duplicates
import pandas as pd
df=pd.DataFrame({'Name':['Alice','Bob','Bob','Charlie','David','David'],'Age':[25,30,30,35,40,40]})
print("Original data:\n",df)
#remove duplicates
df_exact=df.drop_duplicates()
print("After removing duplicates:\n",df_exact)
```

```python
#subset based removal or deletion of duplicates
df=pd.DataFrame({'ID':[1,2,2,3,4,4,5],'Name':['Alice','Bob','Bob','Charlie','David','David','David'],'Age':[25,30,30,35,40,40,40]})
print("Original data:\n",df)
df_subset_id=df.drop_duplicates(subset='ID')
print("\n After removing duplicates(ID): \n",df_subset_id)
df_subset_name=df.drop_duplicates(subset='Name')
print("\n After removing duplicates(Name): \n",df_subset_name)
```

```python
#correcting inconsistent formats
df=pd.DataFrame({'Date':['2025-01-05','05/01/2025','Jan 5, 2025','2025.01.05','2026-08-29','05/01/2026']})
print("Original data:\n",df)
df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
#Case normalization (lower/upper)
df=pd.DataFrame({'Name':['Alice','Bob','Charlie','David']})
df['Name_lower']=df['Name'].str.lower()
df['Name_upper']=df['Name'].str.upper()
print(df)
```
